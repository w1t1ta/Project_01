```python
import torch
from torchvision import models, transforms
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# --- 1) โหลด DeepLabV3+ สำหรับตัดพื้นหลัง ---
def load_deeplabv3_model():
    model = models.segmentation.deeplabv3_resnet101(pretrained=True)
    model.eval()
    return model

# --- 2) ฟังก์ชันตัดพื้นหน้าออก เหลือแต่พื้นหลัง ---
def segment_background(image_path, model):
    image = Image.open(image_path).convert("RGB")
    preprocess = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
    ])
    input_tensor = preprocess(image).unsqueeze(0)

    with torch.no_grad():
        output = model(input_tensor)['out'][0]
        output_predictions = output.argmax(0)
        background_mask = output_predictions == 0  # pixel=0 คือ background

    # เอา foreground ออก
    background_image = np.array(image)
    background_image[~background_mask.numpy()] = 0
    return Image.fromarray(background_image)

# --- 3) Siamese feature extractor (ResNet50 backbone) ---
def load_siamese_resnet():
    resnet = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
    resnet.fc = torch.nn.Identity()
    resnet.eval()
    return resnet

# --- 4) Preprocess รูปให้เข้า ResNet50 ได้ ---
def preprocess_for_resnet(image):
    preprocess = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406],
                             std=[0.229, 0.224, 0.225])
    ])
    return preprocess(image).unsqueeze(0)  # เพิ่ม batch dim

# --- 5) ดึง feature vector ---
def extract_features(image_path, deeplab_model, resnet_model):
    # 1. โหลดภาพ
    image = Image.open(image_path).convert("RGB")
    original_size = image.size  # (W, H)

    # 2. เตรียม input tensor สำหรับ Deeplab
    preprocess_deeplab = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406],
                             std=[0.229, 0.224, 0.225])
    ])
    input_tensor = preprocess_deeplab(image).unsqueeze(0)  # [1, 3, H, W]

    # 3. สร้าง background mask
    with torch.no_grad():
        output = deeplab_model(input_tensor)['out'][0]
        pred_mask = output.argmax(0)  # [H, W]
        bg_mask = (pred_mask == 0).float().unsqueeze(0)  # [1, H, W]

    # 4. Resize mask และภาพให้เข้า ResNet50 (224x224)
    resnet_size = 224
    resize = transforms.Resize((resnet_size, resnet_size))
    resized_mask = resize(bg_mask).unsqueeze(0)  # [1, 1, 224, 224]

    # 5. เตรียมภาพให้เข้า ResNet50
    preprocess_resnet = transforms.Compose([
        transforms.Resize((resnet_size, resnet_size)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406],
                             std=[0.229, 0.224, 0.225])
    ])
    input_for_resnet = preprocess_resnet(image).unsqueeze(0)  # [1, 3, 224, 224]

    # 6. Extract Feature Map จาก ResNet (ก่อน avgpool)
    modules = list(resnet_model.children())[:-2]  # ตัด fc และ avgpool ออก
    backbone = torch.nn.Sequential(*modules)
    with torch.no_grad():
        feature_map = backbone(input_for_resnet)  # [1, 2048, 7, 7]

    # 7. Resize mask ให้ตรงกับ feature map (7x7)
    mask_for_pool = torch.nn.functional.interpolate(resized_mask, size=feature_map.shape[-2:], mode='bilinear')
    mask_for_pool = mask_for_pool.squeeze(1)  # [1, 7, 7]

    # 8. ทำ Masked Average Pooling
    masked_feat = feature_map * mask_for_pool.unsqueeze(1)  # [1, 2048, 7, 7]
    sum_feat = masked_feat.sum(dim=[2, 3])  # [1, 2048]
    norm = mask_for_pool.sum(dim=[1, 2]).clamp(min=1e-6).unsqueeze(1)  # [1, 1]
    avg_feat = sum_feat / norm  # [1, 2048]

    return avg_feat.squeeze(0)  # [2048]

def visualize_background_only(image_path, deeplab_model):
    image = Image.open(image_path).convert("RGB")
    preprocess = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406],
                             std=[0.229, 0.224, 0.225])
    ])
    input_tensor = preprocess(image).unsqueeze(0)

    with torch.no_grad():
        output = deeplab_model(input_tensor)['out'][0]
        pred_mask = output.argmax(0)
        bg_mask = (pred_mask == 0)

    np_img = np.array(image)
    np_img[~bg_mask.numpy()] = 0
    return Image.fromarray(np_img)

# --- 6) คำนวณ Hamming Distance (ตัดที่ Mean) ---
def compute_hamming_distance(vec1, vec2):
    # ตัดที่ค่า mean แทนที่จะตัดที่ 0
    threshold1 = vec1.mean()
    threshold2 = vec2.mean()
    bin_vec1 = (vec1 > threshold1).int()
    bin_vec2 = (vec2 > threshold2).int()
    return (bin_vec1 != bin_vec2).sum().item()

#คำนวณเปอร์เซ็นต์ความแตกต่าง ---
def compute_percentage_difference(hamming_distance, total_bits=2048):
    return (hamming_distance / total_bits) * 100

#คำนวณเปอร์เซ็นต์ความคล้ายคลึง ---
def compute_similarity(percentage_diff):
    return 100 - percentage_diff

#Main เริ่มทำงาน

# Path ภาพ
image_path1 = "/content/484306100_1607547476614172_7718928177540118125_n.jpg"
image_path2 = "/content/484780875_954835580188785_170552539904722165_n.jpg"

# โหลดโมเดล
deeplab_model = load_deeplabv3_model()
siamese_model = load_siamese_resnet()

# ใช้ DeepLab ตัด background
bg1 = segment_background(image_path1, deeplab_model)
bg2 = segment_background(image_path2, deeplab_model)

# ดึง feature vector จาก Siamese ResNet50
feat1 = extract_features(image_path1, deeplab_model, siamese_model)
feat2 = extract_features(image_path2, deeplab_model, siamese_model)

# คำนวณ Hamming Distance
hamming_dist = compute_hamming_distance(feat1, feat2)

# คำนวณเปอร์เซ็นต์ความแตกต่าง
percentage_diff = compute_percentage_difference(hamming_dist)

# คำนวณเปอร์เซ็นต์ความคล้ายคลึง
similarity = compute_similarity(percentage_diff)

print(f"\nHamming Distance ระหว่าง background ที่ได้จาก deeplabv3+: {hamming_dist}")
print(f"Difference: {percentage_diff:.2f}%")
print(f"Similarity: {similarity:.2f}%")

from sklearn.metrics.pairwise import cosine_similarity

# รายการ pairs ที่ต้องการเปรียบเทียบ
pairs = [(0, 1)]  # เปรียบเทียบภาพที่ 1 กับ 2
for i, j in pairs:

    # Cosine Similarity
    cos_sim = cosine_similarity(features[i].unsqueeze(0), features[j].unsqueeze(0))[0][0]
    cos_sim_percent = cos_sim * 100

    print(f"\nCosine Similarity    : {cos_sim:.4f} ({cos_sim_percent:.2f}%)")

# แสดงภาพ background
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(bg1)
plt.title("Background Only - Image 1")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(bg2)
plt.title("Background Only - Image 2")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import torch
from torchvision import models, transforms
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# --- 1) โหลด DeepLabV3+ สำหรับตัดพื้นหลัง ---
def load_deeplabv3_model():
    model = models.segmentation.deeplabv3_resnet101(pretrained=True)
    model.eval()
    return model

# --- 2) ฟังก์ชันตัดพื้นหน้าออก เหลือแต่พื้นหลัง ---
def segment_background(image_path, model):
    image = Image.open(image_path).convert("RGB")
    preprocess = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
    ])
    input_tensor = preprocess(image).unsqueeze(0)

    with torch.no_grad():
        output = model(input_tensor)['out'][0]
        output_predictions = output.argmax(0)
        background_mask = output_predictions == 0  # pixel=0 คือ background

    # เอา foreground ออก
    background_image = np.array(image)
    background_image[~background_mask.numpy()] = 0
    return Image.fromarray(background_image)

# --- 3) Siamese feature extractor (ResNet50 backbone) ---
def load_siamese_resnet():
    resnet = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
    resnet.fc = torch.nn.Identity()  # ลบ fully-connected layer ออก เอาแค่ feature
    resnet.eval()
    return resnet

# --- 4) Preprocess รูปให้เข้า ResNet50 ได้ ---
def preprocess_for_resnet(image):
    preprocess = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406],
                             std=[0.229, 0.224, 0.225])
    ])
    return preprocess(image).unsqueeze(0)  # เพิ่ม batch dim

# --- 5) ดึง feature vector ---
def extract_features(image_path, deeplab_model, resnet_model):
    # 1. โหลดภาพ
    image = Image.open(image_path).convert("RGB")
    original_size = image.size  # (W, H)

    # 2. เตรียม input tensor สำหรับ Deeplab
    preprocess_deeplab = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406],
                             std=[0.229, 0.224, 0.225])
    ])
    input_tensor = preprocess_deeplab(image).unsqueeze(0)  # [1, 3, H, W]

    # 3. สร้าง background mask
    with torch.no_grad():
        output = deeplab_model(input_tensor)['out'][0]
        pred_mask = output.argmax(0)  # [H, W]
        bg_mask = (pred_mask == 0).float().unsqueeze(0)  # [1, H, W]

    # 4. Resize mask และภาพให้เข้า ResNet50 (224x224)
    resnet_size = 224
    resize = transforms.Resize((resnet_size, resnet_size))
    resized_mask = resize(bg_mask).unsqueeze(0)  # [1, 1, 224, 224]

    # 5. เตรียมภาพให้เข้า ResNet50
    preprocess_resnet = transforms.Compose([
        transforms.Resize((resnet_size, resnet_size)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406],
                             std=[0.229, 0.224, 0.225])
    ])
    input_for_resnet = preprocess_resnet(image).unsqueeze(0)  # [1, 3, 224, 224]

    # 6. Extract Feature Map จาก ResNet (ก่อน avgpool)
    modules = list(resnet_model.children())[:-2]  # ตัด fc และ avgpool ออก
    backbone = torch.nn.Sequential(*modules)
    with torch.no_grad():
        feature_map = backbone(input_for_resnet)  # [1, 2048, 7, 7]

    # 7. Resize mask ให้ตรงกับ feature map (7x7)
    mask_for_pool = torch.nn.functional.interpolate(resized_mask, size=feature_map.shape[-2:], mode='bilinear')
    mask_for_pool = mask_for_pool.squeeze(1)  # [1, 7, 7]

    # 8. ทำ Masked Average Pooling
    masked_feat = feature_map * mask_for_pool.unsqueeze(1)  # [1, 2048, 7, 7]
    sum_feat = masked_feat.sum(dim=[2, 3])  # [1, 2048]
    norm = mask_for_pool.sum(dim=[1, 2]).clamp(min=1e-6).unsqueeze(1)  # [1, 1]
    avg_feat = sum_feat / norm  # [1, 2048]

    return avg_feat.squeeze(0)  # [2048]

def visualize_background_only(image_path, deeplab_model):
    image = Image.open(image_path).convert("RGB")
    preprocess = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406],
                             std=[0.229, 0.224, 0.225])
    ])
    input_tensor = preprocess(image).unsqueeze(0)

    with torch.no_grad():
        output = deeplab_model(input_tensor)['out'][0]
        pred_mask = output.argmax(0)
        bg_mask = (pred_mask == 0)

    np_img = np.array(image)
    np_img[~bg_mask.numpy()] = 0
    return Image.fromarray(np_img)

# --- 6) คำนวณ Hamming Distance (ตัดที่ Mean) ---
def compute_hamming_distance(vec1, vec2):
    # ตัดที่ค่า mean แทนที่จะตัดที่ 0
    threshold1 = vec1.mean()
    threshold2 = vec2.mean()
    bin_vec1 = (vec1 > threshold1).int()
    bin_vec2 = (vec2 > threshold2).int()
    return (bin_vec1 != bin_vec2).sum().item()

#คำนวณเปอร์เซ็นต์ความแตกต่าง ---
def compute_percentage_difference(hamming_distance, total_bits=2048):
    return (hamming_distance / total_bits) * 100

#คำนวณเปอร์เซ็นต์ความคล้ายคลึง ---
def compute_similarity(percentage_diff):
    return 100 - percentage_diff

#Main เริ่มทำงาน

# Path ของภาพ
image_path1 = "/content/487823474_1676950896285738_1168476996486110904_n.jpg"
image_path2 = "/content/488712060_1739881346601885_8597199732534888390_n.jpg"

# โหลดโมเดล
deeplab_model = load_deeplabv3_model()
siamese_model = load_siamese_resnet()

# ใช้ DeepLab ตัด background
bg1 = segment_background(image_path1, deeplab_model)
bg2 = segment_background(image_path2, deeplab_model)

# ดึง feature vector จาก Siamese ResNet50
feat1 = extract_features(image_path1, deeplab_model, siamese_model)
feat2 = extract_features(image_path2, deeplab_model, siamese_model)

# คำนวณ Hamming Distance
hamming_dist = compute_hamming_distance(feat1, feat2)

# คำนวณเปอร์เซ็นต์ความแตกต่าง
percentage_diff = compute_percentage_difference(hamming_dist)

# คำนวณเปอร์เซ็นต์ความคล้ายคลึง
similarity = compute_similarity(percentage_diff)

print(f"\nHamming Distance ระหว่าง background ที่ได้จาก deeplabv3+: {hamming_dist}")
print(f"Difference: {percentage_diff:.2f}%")
print(f"Similarity: {similarity:.2f}%")

from sklearn.metrics.pairwise import cosine_similarity

# รายการ pairs ที่ต้องการเปรียบเทียบ
pairs = [(0, 1)]  # เปรียบเทียบภาพที่ 1 กับ 2
for i, j in pairs:

    # Cosine Similarity
    cos_sim = cosine_similarity(features[i].unsqueeze(0), features[j].unsqueeze(0))[0][0]
    cos_sim_percent = cos_sim * 100

    print(f"\nCosine Similarity    : {cos_sim:.4f} ({cos_sim_percent:.2f}%)")

# แสดงภาพ background
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(bg1)
plt.title("Background Only - Image 1")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(bg2)
plt.title("Background Only - Image 2")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import torch
from torchvision import models, transforms
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
import itertools
import requests
from io import BytesIO

# โหลดภาพจาก URL
def load_image_from_url(url):
    response = requests.get(url)
    image = Image.open(BytesIO(response.content)).convert("RGB")
    return image

# (1) โหลด DeepLabV3+
def load_deeplabv3_model():
    model = models.segmentation.deeplabv3_resnet101(pretrained=True)
    model.eval()
    return model

# (2) Segment background
def segment_background(image, model):
    preprocess = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
    ])
    input_tensor = preprocess(image).unsqueeze(0)
    with torch.no_grad():
        output = model(input_tensor)['out'][0]
        output_predictions = output.argmax(0)
        background_mask = output_predictions == 0
    background_image = np.array(image)
    background_image[~background_mask.numpy()] = 0
    return Image.fromarray(background_image)

# (3) Siamese ResNet50
def load_siamese_resnet():
    resnet = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
    resnet.fc = torch.nn.Identity()
    resnet.eval()
    return resnet

# (4) ดึง features
def extract_features(image, deeplab_model, resnet_model):
    preprocess_deeplab = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])
    input_tensor = preprocess_deeplab(image).unsqueeze(0)
    with torch.no_grad():
        output = deeplab_model(input_tensor)['out'][0]
        pred_mask = output.argmax(0)
        bg_mask = (pred_mask == 0).float().unsqueeze(0)
    resnet_size = 224
    resize = transforms.Resize((resnet_size, resnet_size))
    resized_mask = resize(bg_mask).unsqueeze(0)
    preprocess_resnet = transforms.Compose([
        transforms.Resize((resnet_size, resnet_size)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])
    input_for_resnet = preprocess_resnet(image).unsqueeze(0)
    modules = list(resnet_model.children())[:-2]
    backbone = torch.nn.Sequential(*modules)
    with torch.no_grad():
        feature_map = backbone(input_for_resnet)
    mask_for_pool = torch.nn.functional.interpolate(resized_mask, size=feature_map.shape[-2:], mode='bilinear')
    mask_for_pool = mask_for_pool.squeeze(1)
    masked_feat = feature_map * mask_for_pool.unsqueeze(1)
    sum_feat = masked_feat.sum(dim=[2, 3])
    norm = mask_for_pool.sum(dim=[1, 2]).clamp(min=1e-6).unsqueeze(1)
    avg_feat = sum_feat / norm
    return avg_feat.squeeze(0)

# (5) Hamming Distance & Percentage
def compute_hamming_distance(vec1, vec2):
    threshold1 = vec1.mean()
    threshold2 = vec2.mean()
    bin_vec1 = (vec1 > threshold1).int()
    bin_vec2 = (vec2 > threshold2).int()
    return (bin_vec1 != bin_vec2).sum().item()

def compute_percentage_difference(hamming_distance, total_bits=2048):
    return (hamming_distance / total_bits) * 100

def compute_similarity(percentage_diff):
    return 100 - percentage_diff

# ---------- MAIN ----------
# URLs ของภาพ
image_urls = [
    "https://th-images.hellomagazine.com/wp-content/uploads/2024/03/04150616/GHXYoKzXQAAxNvR.jpg",  # เปลี่ยน URL ตามที่ต้องการ
    "https://th-images.hellomagazine.com/wp-content/uploads/2024/03/04150610/GHXQi1BW8AoHjpL-scaled.jpg",  # เปลี่ยน URL ตามที่ต้องการ
    "https://f.ptcdn.info/949/081/000/s2azcs2ddjwA5KsDYRLAv-o.jpg",  # เปลี่ยน URL ตามที่ต้องการ
    "https://kpopping.com/documents/f2/1/800/231010-Winter-at-the-2023-The-Fact-Music-Awards-Red-Carpet-documents-1.jpeg?v=6c309",  # เปลี่ยน URL ตามที่ต้องการ
]


# โหลดโมเดล
deeplab_model = load_deeplabv3_model()
siamese_model = load_siamese_resnet()

# Precompute background images and features
backgrounds = []
features = []

for url in image_urls:
    image = load_image_from_url(url)
    bg = segment_background(image, deeplab_model)
    feat = extract_features(image, deeplab_model, siamese_model)
    backgrounds.append(bg)
    features.append(feat)

# Compare all image pairs
print("\nHamming Distance (Images with background only - DeepLabV3+)\n\n")
pairs = list(itertools.combinations(range(4), 2))

from sklearn.metrics.pairwise import cosine_similarity

for i, j in pairs:
    hamming_dist = compute_hamming_distance(features[i], features[j])
    percent_diff = compute_percentage_difference(hamming_dist)
    similarity = compute_similarity(percent_diff)
    print(f"Pair: img{i+1} vs img{j+1}")
    print(f"Hamming Distance: {hamming_dist}")
    print(f"Difference: {percent_diff:.2f}%")
    print(f"Similarity: {similarity:.2f}%\n")

    # Cosine Similarity
    cos_sim = cosine_similarity(features[i].unsqueeze(0), features[j].unsqueeze(0))[0][0]
    cos_sim_percent = cos_sim * 100

    print(f"Cosine Similarity  : {cos_sim:.4f} ({cos_sim_percent:.2f}%)\n")


# Show all background images
plt.figure(figsize=(12, 6))
for idx, bg in enumerate(backgrounds):
    plt.subplot(1, 4, idx+1)
    plt.imshow(bg)
    plt.title(f"Background - img{idx+1}")
    plt.axis('off')
plt.tight_layout()
plt.show()
```
